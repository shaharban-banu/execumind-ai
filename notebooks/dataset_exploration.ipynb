```python
import pandas as pd
from sqlalchemy import text

from database.database import engine

query = """
SELECT *
FROM reviews
"""

df = pd.read_sql(
    text(query),
    engine
)

print("Total Rows :", len(df))
print("Unique Reviews :", df["review_id"].nunique())
print(
    "Duplicate Review IDs :",
    len(df) - df["review_id"].nunique()
)
```

```python
duplicate_reviews = (
    df["review_id"]
    .value_counts()
)

duplicate_reviews = duplicate_reviews[
    duplicate_reviews > 1
]

print(duplicate_reviews.head(10))
```

```python
duplicate_ids = (
    df["review_id"]
    .value_counts()
)

duplicate_ids = duplicate_ids[
    duplicate_ids > 1
].index

duplicate_rows = df[
    df["review_id"].isin(
        duplicate_ids
    )
]

print(
    duplicate_rows[
        ["review_id", "order_id"]
    ].head(20)
)
```
