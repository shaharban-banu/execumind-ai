```python
from database.database import engine
from database.models import (
    Base,
    Dataset,
    DatasetVersion,
    DatasetFile,
    Customer,
    Order,
    Payment,
    Delivery,
    Product,
    Seller,
    Review,
    OrderItem,
    ExecutiveRecommendation,
)

business_tables = [
    Dataset.__table__,
    DatasetVersion.__table__,
    DatasetFile.__table__,
    Customer.__table__,
    Order.__table__,
    Payment.__table__,
    Delivery.__table__,
    Product.__table__,
    Seller.__table__,
    Review.__table__,
    OrderItem.__table__,
    ExecutiveRecommendation.__table__,
]

print("Dropping business tables...")
Base.metadata.drop_all(bind=engine, tables=business_tables)

print("Creating business tables...")
Base.metadata.create_all(bind=engine, tables=business_tables)

print("Business database reset completed.")
```

```python
import os
from dotenv import load_dotenv

from database.database import SessionLocal
from database.models import User

load_dotenv()

db = SessionLocal()

try:
    username = os.getenv("ADMIN_USERNAME")
    email = "admin@example.com"
    hashed_password = os.getenv("ADMIN_PASSWORD")

    existing = db.query(User).filter(User.username == username).first()

    if existing:
        print("Admin user already exists.")
    else:
        admin = User(
            username=username,
            email=email,
            password_hash=hashed_password,
            role="admin",
            is_active=True,
        )

        db.add(admin)
        db.commit()

        print("Admin user created successfully.")

finally:
    db.close()
```
